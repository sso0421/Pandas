```python
import pandas as pd

pd.set_option("display.max_columns", None)
pd.set_option("display.float_format", "{:,.2f}".format)
```

```python
df = pd.read_csv("kakao_261012.csv", encoding="utf-8-sig")

df.head()
```

```python
# 해당 보고서가 어떻게 생겼는지 = 결측자료, 어떤 컬럼으로 구성, 각 컬럼은 어떤 데이터 타입, 평균, 최소, 최대, 이상치 등
# EDA = Exploratory Data Analysis

df.shape # 61행, 41열 데이터 구성
```

```python
df.columns # 어떤 열로 구성
```

```python
# 아무리 유용한 데이터가 있어도, 연산 및 집계처리가 불가능하다면 무용지물
# 결측 데이터가 존재 => 얼만큼
# 각 컬럼들이 어떤 데이터 타입 구성

df.info()
```

```python
# 현재 데이터의 요약본

df.describe()
```

```python
df = df[df["캠페인 이름"].notna()].copy()
```

```python
# df.head()
# df.shape
df.info()
```

```python
# 광고보고서 세부적으로 분석
# 필요한 컬럼만 찾아올 계획

columns = [
    "캠페인 이름",
    "광고 세트 이름",
    "광고 이름",
    "연령",
    "성",
    "도달",
    "노출",
    "빈도",
    "결과",
    "지출 금액 (KRW)",
    "링크 클릭",
    "클릭(전체)"
]

ads = df[columns].copy()

ads.head()
```

```python
ads.info()
```

```python
ads["결과"] = ads["결과"].fillna(0)
ads["링크 클릭"] = ads["링크 클릭"].fillna(0)
```

```python
# ads.info()
ads.isna().sum()
```

```python
ads["광고 이름"].head() # 2개의 열을 신규 추가!!
```

```python
ads["광고 유형"] = ads["광고 이름"].str.split("_").str[0]
ads["메세지 유형"] = ads["광고 이름"].str.split("_").str[1]
```

```python
ads.head()
```

```python
spend = ads["지출 금액 (KRW)"].sum() # 총 광고비
impressions = ads["노출"].sum() # 총 노출
arrival = ads["도달"].sum() # 총 도달
clicks = ads["링크 클릭"].sum() # 링크 클릭
landing = ads["결과"].sum() # 랜딩 페이지 조회

ctr = clicks / impressions * 100 # CTR
cpc = spend / clicks # CPC
cpm = spend / impressions * 1000 # CPM
cost_per_lading = spend / landing # 랜딩 조회당 비용

print("총 광고비:", f"{spend:,}원")
print("총 노출:", f"{impressions:,}회")
print("총 도달:", f"{arrival:,}회")
print("링크 클릭:", f"{clicks:,.0f}회")
print("랜딩 페이지 조회:", f"{landing:,.0f}회")
print("CTR:", f"{ctr:,.2f}%")
print("CPC:", f"{cpc:,.2f}원")
print("CPM:", f"{cpm:,.2f}원")
print("랜딩 조회당 비용:", f"{cost_per_lading:,.2f}원")
```

```python
funnel = pd.DataFrame({
    "단계": ["노출", "도달", "링크 클릭", "랜딩 페이지 조회"],
    "사용자 행동 수": [impressions, arrival, clicks, landing]
})

funnel
```

```python
import plotly.express as px

fig = px.funnel(
    funnel,
    x="사용자 행동 수",
    y="단계",
    title="Meta 광고 유입 퍼널"
)

fig.update_traces(texttemplate="%{x:,}회") # % = 입력자료형, 외부의 값을 가져와서 해당 값에 어떤 문자열처리를 하고자 할 때, %

fig.show()
```

```python
# 퍼포먼스 마케터
# 1) 과연 해당 캠페인은 전체적으로 성공적이었을까? Yes
# 2) 만약 성공적이라고 판단한다면, 해당 성공에 광고 소재 유형 중 어떤 것이 많은 기여를 했을까?

ads.head()
```

```python
ads["캠페인 이름"] = ads["캠페인 이름"].replace({
    "핑퐁 위시리스트": "핑퐁 위시리스트 - 영상"
})
```

```python
ads.head()
```

```python
campaign = ads.groupby("캠페인 이름", as_index=False).agg({
    "지출 금액 (KRW)": "sum",
    "노출": "sum",
    "링크 클릭": "sum",
    "결과": "sum"
})

campaign["CTR"] = campaign["링크 클릭"] / campaign["노출"] * 100

campaign["CPC"] = campaign["지출 금액 (KRW)"] / campaign["링크 클릭"]

campaign

# 총 예산 : 16,362원
# 영상 : 8,180원
# 이미지 : 8,182원
# 18,000원 : 16,362원 + 1,638원
# 카드정보 입력 : 예산 소진 : 18,000원 + 1,800원
# 선충전 : 16,362원 + 1,638원
```

```python
fig = px.bar(
    campaign,
    x="캠페인 이름",
    y="지출 금액 (KRW)",
    color="캠페인 이름",
    text_auto=True,
    title="캠페인별 광고비 비교"
)

fig.update_traces(texttemplate="%{y:,.0f}원")

fig.show()
```

```python
fig = px.bar(
    campaign,
    x="캠페인 이름",
    y="CTR",
    color="캠페인 이름",
    text_auto=".2f",
    title="캠페인별 CTR 비교"
)

fig.update_traces(texttemplate="%{y:.2f}%")

fig.show()
```

```python
# 퍼포먼스 마케터
# 1) 과연 해당 캠페인은 전체적으로 성공적이었을까? Yes
# 2) 만약 성공적이라고 판단한다면, 해당 성공에 광고 소재 유형 중 어떤 것이 많은 기여를 했을까? 영상
# 3) 영상 > 이미지 클릭을 더 많이 유도 인정 -> 영상 - 혜택 90%, 영상 - 감성 10% // 이미지 - 혜택 50%, 이미지 - 감성 50%
```

```python
creative = ads.groupby("광고 이름", as_index=False)[
    ["지출 금액 (KRW)", "노출", "링크 클릭", "결과"]
].sum()

creative["CTR"] = creative["링크 클릭"] / creative["노출"] * 100

creative["CPC"] = creative["지출 금액 (KRW)"] / creative["링크 클릭"]

creative["LPV"] = creative["지출 금액 (KRW)"] / creative["결과"]

# Landing Page View 비용

creative

# 혜택강조 메세지 : 영상
# 감성자극 메세지 : 이미지
# 직관적인 혜택 : 짧고 굵은 영상 메세지 // 감성 이미지 방식
```
