```python
*** 아나콘다 프로그램 사용 이유!
*저장은 로컬에 진행이 되고 있지만, 실행은 클라우드에서 되어짐
-> 로컬 기반이긴 하나 100%는 아님. (*쨌든? 로컬 프로그램으로 봄)

1. 학습 목적 및 목표 :  파이썬 프로그래밍 언어를 활용해 데이터 수집/분석을 하기 위함
2. 파이썬이라는 프로그래밍 언어는, 다른 언어보다 학습 속도/효과가 빠르기 때문에 배움(난이도 하)
3. 파이썬을 다룰 수 있는 환경을 2가지로 나눠본다면 
- 클라우드(온라인 서버 내 가상 환경) 기반 서비스 ex. 코랩 - 설치가 필요 없으나, 통신의 영향이 多

- 로컬(컴퓨터) 기반 서비스 ex. 아나콘다 - 컴퓨터에 프로그램을 설치하여 안정적으로 파일 관리 
> Web 에디터(ex. VSC) : 주로 프론트엔드(시각적 요소) 개발 시 주 사용 (*내가 작성한 코드가 실시간 웹 구현) 
-> DB, 백엔드 관련 코드 작성하려면 extension이 필수
-> 파이썬의 경우 버전이 정말 다양한데, 각 버전마다 실행이 안 되는 것이 있음 그러면 필요시 매번 가상환경(venv)을 구현해야 함 => 번거로움

> IDE : 통합개발환경
-> 웹 에디터처럼 굳이 프로그래밍 버전 및 운영체제 등의 환경에 따라 별도의 확장 플러그인이나 가상환경을 설치/세팅하지 않아도 됨
-> 즉, 통합개발환경만으로도 모든 것을 구현할 수 있도록 만들어진 프로그램
-> 파이썬 전용 통합개발환경 프로그램으로 가장 적절한 프로그램(무료, 버전관리 및 불필요한 가상환경이 필요 없고, 로컬 설치도 가능)이 아나콘다

*그러면 커서 AI는? 
-> 본인들은 IDE라고 하지만, 사실 설치를 해야 하는 에디터가 맞음. 실제로 설치해야하는 프로그램이 많기 때문
```

```python
*** 실무에서 Pandas의 사용목적

1. 테이블형 데이터를 다루는데 있어서 압도적인 기능
2. 원본데이터의 전처리(Preprocessing) 및 정제(Cleannsing) 작업 시, 비교적 간단한 코드만으로 가능
*정제 : 누락된 값(NaN, Null)을 삭제하거나 평균값/최빈값 등으로 대체
*전처리 : 정제 작업을 완료한 후, 데이터를 알고리즘이나 분석 목적에 맞추는 다음 단계들이 포함
3. 데이터 분석 기법을 활용할 때 굉장히 많이 등장함

- EDA : 탐색적 데이터 분석 = Exploratory Data Analysis
> 본격적인 데이터 분석 전 데이터의 사이즈, 결측치, 이상치, 평균 및 중앙값 사분위 데이터 등을 통해서 데이터 패턴을 파악하기 위한 목적을 가지고 있음
    
- LDA : 잠재적 주제도출 분석 = Latent Dirichlet Allocation = 잠재 디리클레 할당
> EDA 과정을 거쳐서 충분히 분석할만한 가치가 있다고 판단은 하였으나, 파편화 되어있는 데이터들이 어떤 주제로 구성되어 있는지를 분류하고 파악하기 위한 목적
> 사용자 및 SNS 키워드 및 리뷰 수집 결과 수가 100만개라고 한다면, 중복되어지는 내용이 많음(나쁘다, 안 좋다, 불편하다) -> 이러한 것을 LDA 과정에서 처리 해야 함

- ETL : 데이터 추출처리 분석 = Extract Transform Load
> 원본 데이터 = 원시데이터는 정보(데이터의 패턴조회, 가설설정)가 아님 -> 원시데이터를 유용한 정보성 요소로 변환시키도록하기 위한 분석 과정
```
