```python
# 간단한 실습!
# 기본 개념 = 확률

#주사위 → 반드시 1개의 면 상단부 : 만약 3이 나올 확률은 ? : 1/6 표기를 P(x=3)으로 표기함

# P(X=3)=1/6 : 6개의 후보군 가운데, 3이라는 x변수 값에 들어가 있는 해당 숫자가 나올 수 있는 확률은 1/6
# P => 함수 : "P"robaility Measure = 확률 측도
# X => 확률변수 : X = 3 : 3이라는 값이  나올 사건 (event)

# P(사건) : 해당 사건이 발생할 확률 = 1/6
```

```python
# 1. 조건부 확률
# 2. 결합 확률
# 3. 주변 확률
```

```python
# 1. 조건부 확률 : 어떤 사건이 일어났다는 조건 하에 다른 사건이 일어날 확률
# -> P(나타날 조건|먼저 정해진 조건을 작성)
Q. 상자 A에서 딸기맛 사탕이 뽑힐 확률
*조건부 확률 표기법 : P(Y=딸기맛|X=A) == P(딸기맛|A) == 2/10
=> A상자에서 딸기맛 사탕 개수 / A상자 전체 상태 개수
```

```python
# 2. 결합 확률 : 두 가지 사건이 동시에 일어날 확률
# 상자 A가 선택되고, 상자 A에서 딸기맛 사탕이 뽑힐 확률
# -> P(조건1, 조건2)
Q. A가 뽑히고 딸기맛이 뽑힐 확률
P(조건1, 조건2) = P(조건1) X P(조건2)
=> P(A, 딸기맛) = P(A)xP(딸기맛|A) == 6/10 x 2/10 = 12/100
```

```python
--- 개념 ---
# 라이브러리 모듈 : skit-learn
# 복잡한 연산 및 알고리즘을 직접 구현하지 않아도 메서드, 속성만으로도 실행할 수 있도록 만들어진 라이브러리 모듈
# 확률 기반의 패턴인식 샘플 데이터를 제공
# 어떤 문제 데이터 세트 학습 -> 머신러닝 모델에게 학습 -> 또 다른 (실제) 문제와 정답을 제공 (정확도 %를 도출)
# 원천 소스(pattern source = raw data)
# 머신러닝이 패턴을 찾아내고
# 부류(방) = class
# 분류 = clssification (원천 소스 내 패턴을 학습한 후, 비슷한 것들끼리 모아놓을 수 있도록 사전에 만들어 놓은 부류에 분류해서 값을 담아놓는 행위를 할 것 = 지도학습)
# 숫자 = 직접 사람이 작성한 글씨를 기반(원천소스)

--- 실전/실무 ---
# 1) 샘플(로우 데이터)가 필요함
# 2) 샘플 데이터를 2개로 나누어야 함
# - training set(훈련집합) : 정답 o
# - test set(테스트집합) : 정답 x
# 3) 분류기 설계 = 모델 선정 ex) RF, XG Boost,, (*현재 우리가 해결하고자 하는 문제에 가장 적합한 모델을 선정)
# 4) 해당 모델에게 학습시킬 수 있는 형태로 자료를 변경 (전처리, 정제 가공 => feature engineering *단, 문자를 읽을 수 없고 수치화 해야 함)
# 5) 훈련집 데이터 세트를 활용하여 분류기에게 학습 시킴
# 6) 분류기 = 모델, 테스트집합 세트를 주고 정답을 맞혀라! 라고 명령
```
