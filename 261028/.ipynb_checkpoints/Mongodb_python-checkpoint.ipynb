```python
import requests
from bs4 import BeautifulSoup
from pymongo import MongoClient

client = MongoClient("mongodb://localhost:27017/")
db = client["giftshop"]
collection = db["teddyproducts"]

# hostname : 서버에 접속하기 위한 주소 : localhost = 
# port : 해당 서버 주소 내 독립되어있는 공간 : 3306, 8000, 27017

documents = list()

for page_num in range(9) :

    if page_num == 0 :
        res = requests.get("https://davelee-fun.github.io/")
    
    else:
        res = requests.get(f"https://davelee-fun.github.io/page{page_num + 1}")
        
    soup = BeautifulSoup(res.text, "html.parser")
    data = soup.select("div.card-body")
    
    for item in data : 
        
        category = item.select_one("h2.card-title").get_text().replace("관련 상품 추천", "").strip()
        product = item.select_one("h4.card-text").get_text().replace("상품명:", "").strip()

        doc = {
            "title": product,
            "category": category
        }

        documents.append(doc)
    
if documents :
    collection.insert_many(documents)

client.close()

print("mongodb 데이터 전송 완료!")
```
